# Task 8: Check Missing Values

## Objective

Real-world datasets often contain missing values.

Before building a Machine Learning model, it is important to identify the number of missing values in each feature.

This helps us decide whether to:
- Remove the feature
- Fill missing values using statistical techniques
- Keep the feature as it is

In this step, we calculate the total number and percentage of missing values in every column.

```python
# Count missing values
missing_values = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (missing_values / len(df)) * 100

# Create a summary table
missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})

# Display columns having missing values
missing_df = missing_df[missing_df["Missing Values"] > 0]

# Sort in descending order
missing_df = missing_df.sort_values(by="Percentage", ascending=False)

print("Total Columns with Missing Values:", missing_df.shape[0])

missing_df.head(20)
```

# Task 9: Visualize Missing Values

## Objective

After identifying the missing values in the dataset, it is important to visualize their distribution.

Instead of plotting a heatmap for all 592 features (which becomes difficult to interpret), we visualize the **Top 30 features with the highest percentage of missing values**.

This helps us identify the features that may need to be removed or imputed during data preprocessing.

```python
# Calculate missing percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Keep only columns with missing values
missing_percentage = missing_percentage[missing_percentage > 0]

# Sort in descending order
missing_percentage = missing_percentage.sort_values(ascending=False)

# Plot Top 30 features
plt.figure(figsize=(16,6))

missing_percentage.head(30).plot(
    kind='bar',
    color='steelblue',
    edgecolor='black'
)

plt.title("Top 30 Features with Highest Missing Percentage", fontsize=15)
plt.xlabel("Features")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=90)

plt.show()
```

# Task 10: Remove Highly Missing Features

## Objective

Features containing a very high percentage of missing values provide little useful information for prediction.

To improve model performance and reduce noise, columns having more than **70% missing values** are removed from the dataset.

This step also reduces computational complexity during model training.

```python
# ===========================================
# Task 10 : Remove Highly Missing Features
# ===========================================

# Threshold
threshold = 70

# Identify columns
columns_to_drop = missing_percentage[
    missing_percentage > threshold
].index

print("Number of Columns to Drop :", len(columns_to_drop))

# Drop columns
df = df.drop(columns=columns_to_drop)

print("\nDataset Shape After Dropping Columns")
print(df.shape)
```

# Task 12: Missing Value Imputation

## Objective

Remaining missing values are replaced using the **median** of each numerical feature.

# Task 29: Remove Low Variance Features

## Objective

Some sensor features may have constant or nearly constant values across all observations.

Such features contribute very little to the prediction process and unnecessarily increase the dimensionality of the dataset.

In this step, features with zero variance are identified and removed using the Variance Threshold technique.

```python
from sklearn.feature_selection import VarianceThreshold

# Separate only predictor variables
X = df.drop(columns=["Pass/Fail", "Time"])

print("Original Number of Features :", X.shape[1])

# Create Variance Threshold object
selector = VarianceThreshold(threshold=0)

# Fit selector
selector.fit(X)

# Keep only selected columns
selected_columns = X.columns[selector.get_support()]

# Create new dataframe with selected features
X = X[selected_columns]

print("Number of Features After Removing Constant Features :", X.shape[1])
```

# Task 30: Separate Predictor and Target Variables

## Objective

Machine Learning models require the dataset to be divided into:

- **Predictor Variables (X):** Sensor measurements and engineered features used as input.
- **Target Variable (y):** Pass/Fail status of the semiconductor manufacturing process.

The predictors are stored in **X**, while the target variable is stored in **y**.

```python
# Target variable
y = df["Pass/Fail"]

print("Predictor Shape :", X.shape)
print("Target Shape    :", y.shape)

print("\nTarget Class Distribution:")
print(y.value_counts())
```

# Task 31: Split the Dataset into Training and Testing Sets

## Objective

The dataset is divided into training and testing sets.

- Training data is used to train the machine learning models.
- Testing data is used to evaluate model performance on unseen observations.

An 80:20 split is used in this project.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set :", X_train.shape)
print("Testing Set :", X_test.shape)
```

# Task 32: Feature Standardization

## Objective

Machine Learning algorithms such as Logistic Regression, Support Vector Machine (SVM), and K-Nearest Neighbors (KNN) are sensitive to the scale of input features.

Feature standardization transforms the predictor variables so that they have:

- Mean = 0
- Standard Deviation = 1

The StandardScaler is fitted using only the training data and then applied to both the training and testing datasets. This prevents data leakage and ensures fair model evaluation.

```python
from sklearn.preprocessing import StandardScaler

# Initialize StandardScaler
scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data
X_test_scaled = scaler.transform(X_test)

print("Feature Standardization Completed Successfully!")

print("\nTraining Data Shape :", X_train_scaled.shape)
print("Testing Data Shape  :", X_test_scaled.shape)
```

# Task 33: Analyze Class Distribution Before SMOTE

## Objective

Before applying any balancing technique, it is important to examine the distribution of the target classes in the training dataset.

If one class significantly outnumbers the other, the dataset is considered imbalanced. Such imbalance may bias machine learning models toward the majority class.

The class distribution is visualized using a count plot.

```python
print("Training Class Distribution:\n")
print(y_train.value_counts())

plt.figure(figsize=(6,5))

ax = sns.countplot(
    x=y_train,
    color="steelblue"
)

# Add value labels
for p in ax.patches:
    ax.annotate(
        str(int(p.get_height())),
        (p.get_x()+p.get_width()/2, p.get_height()),
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.title("Training Class Distribution Before SMOTE")
plt.xlabel("Pass / Fail")
plt.ylabel("Count")

plt.show()
```

# Task 34: Apply SMOTE

## Objective

The training dataset is highly imbalanced, with significantly fewer failure samples than pass samples.

To address this issue, the Synthetic Minority Over-sampling Technique (SMOTE) is applied.

SMOTE generates synthetic samples for the minority class, resulting in a balanced training dataset. This helps machine learning models learn both classes more effectively and reduces bias toward the majority class.

```python
from imblearn.over_sampling import SMOTE

# Initialize SMOTE
smote = SMOTE(random_state=42)

# Apply SMOTE only on training data
X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Before SMOTE:\n")
print(y_train.value_counts())

print("\nAfter SMOTE:\n")
print(y_train_smote.value_counts())
```

# Task 35: Verify Balanced Dataset

## Objective

After applying SMOTE, the class distribution is visualized to confirm that the training dataset is balanced.

A balanced dataset helps improve the ability of machine learning models to correctly classify both majority and minority classes.

```python
plt.figure(figsize=(6,5))

ax = sns.countplot(
    x=y_train_smote,
    color="seagreen"
)

# Add labels
for p in ax.patches:
    ax.annotate(
        str(int(p.get_height())),
        (p.get_x()+p.get_width()/2, p.get_height()),
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.title("Training Class Distribution After SMOTE")
plt.xlabel("Pass / Fail")
plt.ylabel("Count")

plt.show()
```

# Task 36: Logistic Regression Model

## Objective

Logistic Regression is one of the most widely used supervised learning algorithms for binary classification problems.

In this task, a Logistic Regression model is trained using the balanced training dataset generated after applying SMOTE.

The model is then evaluated on the testing dataset to determine its predictive performance.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Create model
lr = LogisticRegression(random_state=42, max_iter=1000)

# Train model
lr.fit(X_train_smote, y_train_smote)

# Prediction
y_pred_lr = lr.predict(X_test_scaled)

# Accuracy
lr_accuracy = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Accuracy : {:.4f}".format(lr_accuracy))

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_lr))
```

GridSearchCV evaluates multiple combinations of hyperparameters using cross-validation and selects the combination that produces the highest classification accuracy.

This helps improve the predictive performance of the final model.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Parameter grid
param_grid = {
    "C": [0.1, 1, 10],
    "kernel": ["linear", "rbf"],
    "gamma": ["scale", "auto"]
}

# Grid Search
grid = GridSearchCV(
    estimator=SVC(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid.fit(X_train_smote, y_train_smote)

print("Best Parameters:")
print(grid.best_params_)

print("\nBest Cross Validation Accuracy:")
print(round(grid.best_score_,4))
```
